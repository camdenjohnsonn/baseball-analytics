```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

os.chdir(r"C:\Users\camde\Desktop\baseball_analytics")

df = pd.read_csv('data/processed/batting_cleaned.csv')
```

```python
sns.histplot(df['next_OPS'])
plt.title('Next OPS Distribution')
plt.xlabel('Next OPS')
plt.ylabel('Frequency')
plt.savefig('outputs/figures/next_ops_dist.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
sns.scatterplot(df, x='OPS', y = 'next_OPS')
plt.title('Next OPS vs OPS')
plt.savefig('outputs/figures/next_ops_vs_ops.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
age_ops = df.groupby('age')['OPS'].mean()
sns.lineplot(df, x='age', y = age_ops)
plt.title('Average OPS vs age')
plt.xticks(rotation=45)
plt.savefig('outputs/figures/avg_ops_vs_age.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
sns.scatterplot(df, x='salary', y='OPS')
plt.title('OPS vs Salary')
plt.savefig('outputs/figures/ops_vs_sal.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
sns.scatterplot(df, x='salary', y='OPS')
plt.title('OPS vs log(Salary)')
plt.xscale('log')
plt.savefig('outputs/figures/ops_vs_log_sal.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
corr_matrix = df.select_dtypes(include = 'number').corr()['next_OPS'].sort_values()
corr_matrix.plot(kind='barh')
plt.title('Correlation with next_OPS')
plt.savefig('outputs/figures/corr_w_next_ops.png', dpi=150, bbox_inches='tight')
plt.show()
```
